```python
import pandas as pd

sales = pd.read_csv("salesdata.csv")
```

```python
sales.head()
```

```python
sales.tail()
```

```python
sales.info()
```

```python
sales['OrderDate'] = pd.to_datetime(sales['OrderDate'], format= 'mixed', errors= 'coerce')
```

```python
sales.info()
```

```python
sales['UnitPrice'].unique()
```

```python
sales['UnitPrice'] = sales['UnitPrice'].astype (str).replace('$','',regex=False)
sales['UnitPrice'] = pd.to_numeric(sales['UnitPrice'], errors='coerce')
```

```python
sales.info()
```

```python
sales['Discount'] = sales['Discount'].astype (str).replace('%','',regex=False)
sales['Discount'] = pd.to_numeric(sales['Discount'], errors='coerce')
```

```python
sales.info()
```
